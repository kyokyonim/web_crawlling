```python
import requests 
from bs4 import BeautifulSoup
import pandas as pd 

#1. 크롤링할 시작 URL 정의하기 
base_url = "https://search.naver.com/search.naver?sm=tab_hty.top&where=news&ssc=tab.news.all&query=%EC%9B%90%EC%9E%90%EB%A0%A5&oquery=%EC%9B%90%EC%9E%90%EB%A0%A5&tqi=j7lD0wqosTCssfmJ%2FO0ssssss1K-481109&ackey=8ee91f0s"

#2. 기사 제목을 담을 리스트 만들기
news_title = []

#3. 여러 페이지를 수집하기 위해 반복문 사용
for page in range(1,4):
    #페이지 번호에 맞게 start 값 계산(1페이지 -1 2 페이지 - 11 3패이지 -21)
    start_num = (page -1) * 10 + 1
    url = base_url + str(start_num)

    #해당 URL의 HTML 가져옴 
    response = requests.get(url)
    html = response.text

    #HTML 분석하기 쉬운 형태로 만들기
    soup = BeautifulSoup(html, 'html.parser')

    #span 태그이면서 class 가 ''인 요소를 찾는다 (이 부분이 기사 제목이다)
    titles = soup.select('span.sds-comps-text-type-headline1')

    #찾는 제목들을 리스트에 추가
    for title in titles:
        news_title.append(title.get_text())
    
 #수집된 제목들을 확인하기 
print(f"{len(news_title)}개의 가사 제목을 수집했습니다.")
print(news_title[:5]) #앞에 5개만 출력

#수집한 데이터를 DataFrame으로 만들고 CSV 파일로 저장!
df = pd.DataFrame(news_title, columns=['title'])
df.to_csv('nuclear_news_titles.csv', index=False, encoding='utf-8-sig')
```
